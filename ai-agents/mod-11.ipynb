## ReAct Framework

The ReAct (Reason and Act) framework allows AI agents to reason about a task and then act upon it. This involves generating a plan, executing actions, and observing the outcomes to refine future actions. It is particularly useful in dynamic environments where conditions change frequently.

### In-depth Explanation

The ReAct framework is built on the idea that an AI agent should not only perform tasks but also understand and adapt to the environment. The agent reasons about the task, forms a plan, executes actions based on that plan, and observes the outcomes to learn and improve.

### Real-World Case Study

**Autonomous Vehicles:** Self-driving cars use the ReAct framework to navigate complex and dynamic environments. The agent reasons about the current traffic conditions, plans a route, executes driving actions, and observes the environment to make real-time adjustments.

### Hands-On Code Example



**Explanation:**
- We define a target number randomly between 1 and 10.
- The agent repeatedly chooses a new number until it matches the target.
- Each attempt is recorded, and the process demonstrates the reasoning and acting cycle.

```python
import random

def react_agent():
    # Define a simple task: find a random number between 1 and 10
    target = random.randint(1, 10)
    current = 0
    attempts = 0
    
    while current!= target:
        # Reason: Choose a new number
        current = random.randint(1, 10)
        # Act: Increment attempts
        attempts += 1
        print(f'Attempt {attempts}: Current number is {current}')
    
    print(f'Target {target} found in {attempts} attempts!')

# Run the ReAct agent
react_agent()
```

## LangGraph for Multi-Agent Systems

LangGraph is a framework for creating multi-agent systems where multiple AI agents collaborate to solve complex problems. Each agent can have its own goals and actions, and they communicate to achieve a common objective. This is particularly useful in scenarios requiring coordinated efforts among different entities.

### In-depth Explanation

Multi-agent systems are essential in scenarios where a single agent cannot accomplish the task alone. Agents collaborate, share information, and work towards a common goal. This approach is beneficial in environments with high complexity and interdependence.

### Real-World Case Study

**Supply Chain Management:** In a supply chain, multiple agents representing different entities (suppliers, manufacturers, distributors) collaborate to optimize the flow of goods. Each agent has specific goals (e.g., minimizing cost, maximizing efficiency), and they communicate to ensure smooth operations.

### Hands-On Code Example



**Explanation:**
- We define two simple agents with basic actions.
- We create a LangGraph instance and add the agents to it.
- We define an interaction between the agents.
- Running the graph demonstrates how agents can collaborate.

```python
from langgraph import LangGraph

# Define two simple agents
def agent1():
    return 'Agent 1: Hello!'

def agent2():
    return 'Agent 2: Hi there!'

# Create a LangGraph instance
graph = LangGraph()

# Add agents to the graph
graph.add_agent('agent1', agent1)
graph.add_agent('agent2', agent2)

# Define interactions
graph.add_interaction('agent1', 'agent2')

# Run the multi-agent system
output = graph.run()
print(output)
```

- **Real-World Applications:** Autonomous vehicles and supply chain management.
