# Introduction to AI Agents

## 1. What Makes an LLM an "Agent"?

An LLM is just the "brain" of an agent. To become a true agent, the system needs a framework wrapped around the LLM that provides it with:

1. **Memory:** The ability to remember past interactions, steps taken, and lessons learned (short-term and long-term).
2. **Tools (or Actions):** The ability to interact with the outside world. This could be searching the web, executing code, querying a database, or sending an email.
3. **Planning & Reasoning:** A methodology to break a complex goal into smaller sub-tasks and evaluate progress.

### The ReAct Framework

One of the most popular frameworks for agentic behavior is **ReAct (Reasoning and Acting)**. In ReAct, the agent operates in a continuous loop:

```text
┌─────────────────────────────────┐
│  1. Observe Environment         │
│     (Receive user prompt)       │
└────────────┬────────────────────┘
             │
┌────────────▼────────────────────┐
│  2. Think/Reason                │
│     (What do I need to do?)     │
└────────────┬────────────────────┘
             │
┌────────────▼────────────────────┐
│  3. Plan Action                 │
│     (Decide on tool & inputs)   │
└────────────┬────────────────────┘
             │
┌────────────▼────────────────────┐
│  4. Execute Action              │
│     (Run Python, search web)    │
└────────────┬────────────────────┘
             │
┌────────────▼────────────────────┐
│  5. Observe Result              │
│     (Analyze tool output)       │
└────────────┬────────────────────┘
             │
             └──────────┐
                        │ Loop until goal is met
                        ▼
```

## 2. Real-World Case Studies

### Case Study 1: Customer Support Automation

Many companies use AI agents to handle customer support queries. For example, an AI agent can be designed to:

1. **Observe:** Receive a customer's query via chat.
2. **Think/Reason:** Determine if the query is about product information, order status, or a return request.
3. **Plan Action:** Use different tools such as a product database, order management system, or return processing API.
4. **Execute Action:** Fetch the required information or perform the requested action.
5. **Observe Result:** Provide the customer with the relevant information or confirmation.

#### Example: Amazon Customer Support

Amazon uses AI agents to handle a large volume of customer queries. When a customer asks about their order status, the agent retrieves the information from the order management system and provides a timely response. This reduces the workload on human support staff and ensures quick, accurate responses to customers.

### Case Study 2: Financial Trading

AI agents are used in algorithmic trading to make split-second decisions. An agent might:

1. **Observe:** Monitor real-time stock market data.
2. **Think/Reason:** Analyze trends, news sentiment, and historical data.
3. **Plan Action:** Decide whether to buy, sell, or hold a particular stock.
4. **Execute Action:** Place trades through a brokerage API.
5. **Observe Result:** Monitor the performance of the trades and adjust strategies accordingly.

#### Example: Jane Street

Jane Street, a global market maker, uses AI agents to execute trades. These agents analyze vast amounts of data in real-time, make quick decisions, and execute trades with minimal human intervention. This allows them to take advantage of market opportunities that would be impossible for a human to spot and act on in time.

## 3. Hands-On: Building a Simple Agent with LangChain

Let's look at how we construct an agent using Python and the `langchain` library. This script demonstrates how to equip an LLM with a calculator tool so it can solve math problems it would otherwise hallucinate on.

### Python Code Example



**Expected Verbose Output:**
```text
> Entering new AgentExecutor chain...
I need to calculate 15.3 raised to the power of 4.7.
Action: Calculator
Action Input: 15.3 ^ 4.7
Observation: Answer: 357494.484...
Thought: I now know the final answer.
Final Answer: 15.3 raised to the power of 4.7 is approximately 357,494.48.

> Finished chain.
```

As you can see, the agent didn't just guess the answer. It recognized it needed to do math, selected the `Calculator` tool, fed it the correct formula, observed the result, and formulated a final response.

````python
# Note: This requires the langchain and openai packages
# !pip install langchain openai

from langchain_openai import ChatOpenAI
from langchain.agents import load_tools, initialize_agent, AgentType

# 1. Initialize the "Brain" (The LLM)
# We set temperature=0 so the agent is deterministic and analytical
llm = ChatOpenAI(temperature=0, model_name="gpt-3.5-turbo")

# 2. Equip the Agent with Tools
# We give it a calculator tool ('llm-math') to perform accurate calculations
tools = load_tools(["llm-math"], llm=llm)

# 3. Initialize the Agent
# We use the ZERO_SHOT_REACT_DESCRIPTION agent type
# This tells the agent to use the ReAct framework to figure out which tool to use
agent = initialize_agent(
    tools, 
    llm, 
    agent=AgentType.ZERO_SHOT_REACT_DESCRIPTION, 
    verbose=True # Verbose mode lets us see its thought process!
)

# 4. Give the Agent a Goal
question = "What is 15.3 raised to the power of 4.7?"

print(f"Goal: {question}\n")
print("--- Agent Execution Trace ---")

# Run the agent
result = agent.run(question)

print("\n--- Final Output ---")
print(result)
````
