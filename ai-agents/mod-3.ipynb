# LangGraph Basics

## Understanding LangGraph Components

LangGraph is built on top of LangChain and provides a way to define directed acyclic graphs (DAGs) of operations. Each node in the graph represents a step in the workflow, which can be a language model call, a tool invocation, or a conditional branch. Nodes are connected by edges that define the flow of data and control between steps.

### Why Use LangGraph?

LangGraph is particularly useful for:
- **Complex Workflows:** Breaking down complex tasks into manageable steps.
- **State Management:** Maintaining state across multiple steps in a workflow.
- **Conditional Logic:** Implementing dynamic workflows that adapt based on intermediate results.
- **Tool Integration:** Leveraging external tools and APIs within the workflow.

### Basic Components

- **Nodes:** Represent individual steps in the workflow. Each node can perform a specific task, such as calling a language model or executing a tool.
- **Edges:** Define the flow of data and control between nodes. Edges can be conditional, directing the workflow based on the output of previous nodes.

## Creating a Simple Workflow

Let's start with a simple example to illustrate the basic structure of a LangGraph workflow.



**Explanation:**
- We create a `Graph` instance to define our workflow.
- We add two nodes: `start` and `end`. The `start` node returns a greeting message, and the `end` node formats the received message.
- We define an edge from `start` to `end`, indicating the flow of data.
- Running the workflow prints `Received: Hello, LangGraph!`.

```python
from langgraph import Graph

# Create a LangGraph instance
workflow = Graph()

# Define nodes
workflow.add_node('start', lambda: 'Hello, LangGraph!')
workflow.add_node('end', lambda x: f'Received: {x}')

# Define edges
workflow.add_edge('start', 'end')

# Run the workflow
result = workflow.run()
print(result)
```

## Creating Conditional Workflows

LangGraph allows you to create conditional workflows where the next step depends on the output of the previous step. This is achieved using conditional nodes that evaluate the output and direct the flow accordingly.

### Real-World Case Study: Customer Support Chatbot

A customer support chatbot can use LangGraph to handle different user queries dynamically. Based on the user's input, the chatbot can direct the conversation to appropriate handlers (e.g., billing, technical support, general inquiries).



**Explanation:**
- The `start` node initiates the workflow with a message.
- The `condition` node evaluates the message. If the condition is met (`True`), it directs the flow to `true_branch`; otherwise, it goes to `false_branch`.
- Both branches lead to the `end` node, which formats the final output.

```python
from langgraph import Graph

# Create a LangGraph instance
workflow = Graph()

# Define nodes
workflow.add_node('start', lambda: 'Check condition')
workflow.add_node('condition', lambda x: x == 'Check condition')
workflow.add_node('true_branch', lambda: 'Condition is true')
workflow.add_node('false_branch', lambda: 'Condition is false')
workflow.add_node('end', lambda x: f'Final output: {x}')

# Define edges
workflow.add_edge('start', 'condition')
workflow.add_edge('condition', 'true_branch', condition=True)
workflow.add_edge('condition', 'false_branch', condition=False)
workflow.add_edge('true_branch', 'end')
workflow.add_edge('false_branch', 'end')

# Run the workflow
result = workflow.run()
print(result)
```

## Integrating External Tools

LangGraph can integrate external tools and APIs to enhance the capabilities of your workflows. For example, you can call a weather API to get real-time weather data within your workflow.

### Hands-On Example: Weather API Integration



**Explanation:**
- The `start` node initiates the workflow.
- The `fetch_weather` node calls the OpenWeatherMap API to get current weather data for London.
- The `format_weather` node formats the temperature data.
- The `end` node outputs the final result.

```python
import requests
from langgraph import Graph

# Create a LangGraph instance
workflow = Graph()

# Define nodes
workflow.add_node('start', lambda: 'Fetch weather data')
workflow.add_node('fetch_weather', lambda: requests.get('https://api.openweathermap.org/data/2.5/weather?q=London&appid=your_api_key').json())
workflow.add_node('format_weather', lambda data: f"Current temperature in London: {data['main']['temp']} K")
workflow.add_node('end', lambda x: f'Final output: {x}')

# Define edges
workflow.add_edge('start', 'fetch_weather')
workflow.add_edge('fetch_weather', 'format_weather')
workflow.add_edge('format_weather', 'end')

# Run the workflow
result = workflow.run()
print(result)
```
