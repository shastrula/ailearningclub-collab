### Hybrid Search

**Explanation:** Hybrid search combines keyword and semantic searches for optimal results. This approach leverages the strengths of both methods to provide comprehensive search results.

**Real-World Case Study:** Microsoft uses hybrid search in Bing to deliver more accurate and relevant search results by combining keyword matching with semantic understanding.

**Hands-On Code Example:**

````python
import faiss

# Example: Creating a simple vector database using Faiss
dimension = 128  # Dimension of the vectors
index = faiss.IndexFlatL2(dimension)  # L2 distance metric
vectors = [[1.0]*dimension, [2.0]*dimension]  # Sample vectors
index.add(vectors)  # Adding vectors to the index

# Searching for nearest neighbors
query_vector = [1.5]*dimension
distances, indices = index.search([query_vector], 2)  # Searching for 2 nearest neighbors
print(f'Distances: {distances}, Indices: {indices}')
````

````python
import re

def chunk_text(text, chunk_size=500):
    return [text[i:i+chunk_size] for i in range(0, len(text), chunk_size)]

text = "This is a long text that needs to be chunked into smaller pieces for processing."
chunks = chunk_text(text)
print(chunks)
````

````python
def rerank_results(results, relevance_scores):
    ranked_results = sorted(zip(results, relevance_scores), key=lambda x: x[1], reverse=True)
    return [result for result, score in ranked_results]

results = ["result1", "result2", "result3"]
relevance_scores = [0.8, 0.6, 0.9]
reranked_results = rerank_results(results, relevance_scores)
print(reranked_results)
````

````python
from sklearn.feature_extraction.text import TfidfVectorizer
from sentence_transformers import SentenceTransformer

# Keyword search
vectorizer = TfidfVectorizer()
corpus = ["This is the first document.", "This document is the second document."]
tfidf_matrix = vectorizer.fit_transform(corpus)

# Semantic search
model = SentenceTransformer('paraphrase-MiniLM-L6-v2')
embeddings = model.encode(corpus)

def hybrid_search(query, tfidf_matrix, embeddings, vectorizer):
    query_tfidf = vectorizer.transform([query])
    query_embedding = model.encode([query])
    
    # Keyword search
    keyword_scores = query_tfidf.dot(tfidf_matrix.T).toarray().ravel()
    
    # Semantic search
    semantic_scores = embeddings @ query_embedding.T
    
    # Combine scores
    combined_scores = keyword_scores + semantic_scores.ravel()
    ranked_results = sorted(range(len(combined_scores)), key=lambda i: combined_scores[i], reverse=True)
    
    return [corpus[i] for i in ranked_results]

query = "document"
results = hybrid_search(query, tfidf_matrix, embeddings, vectorizer)
print(results)
````
