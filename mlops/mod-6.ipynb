# Model Registry and Model Serving

## Model Registry

### In-depth Explanation

A **Model Registry** is a centralized repository designed to store, version, and manage machine learning models. It serves as a crucial component in the MLOps workflow, enabling data scientists and ML engineers to:

- **Track Model Versions:** Maintain a history of all model versions, allowing for easy comparison and rollback to previous versions if necessary.
- **Compare Performance Metrics:** Evaluate different model versions based on performance metrics such as accuracy, precision, recall, and F1 score.
- **Simplify Deployment:** Streamline the process of deploying models to production by providing a single source of truth for model artifacts.

### Real-World Case Study

**Company:** Netflix  
**Use Case:** Recommendation System  
Netflix uses a Model Registry to manage the numerous machine learning models that power its recommendation system. By versioning and tracking models, Netflix can quickly iterate on improvements, roll back to stable versions in case of issues, and ensure that the recommendation system remains performant and relevant.

### Hands-On Code Example

Below is a Python code example using MLflow to register a machine learning model.



**Explanation:**
- **mlflow.set_tracking_uri("http://localhost:5000")**: Sets the MLflow tracking server URI.
- **mlflow.sklearn.log_model(model, "model")**: Logs the trained model to the MLflow Model Registry.
- **run.info.run_uuid**: Retrieves the unique run ID for the logged model.

```python
import mlflow
import mlflow.sklearn
from sklearn.datasets import load_iris
from sklearn.ensemble import RandomForestClassifier

# Initialize MLflow
mlflow.set_tracking_uri("http://localhost:5000")

# Load data and train a simple model
iris = load_iris()
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(iris.data, iris.target)

# Log the model to MLflow
with mlflow.start_run() as run:
    mlflow.sklearn.log_model(model, "model")
    run_id = run.info.run_uuid
    print(f'Model logged with run ID: {run_id}')
```

## Model Serving

### In-depth Explanation

**Model Serving** refers to the process of deploying machine learning models into a production environment where they can receive input data, make predictions, and return results. Effective model serving ensures that models are:

- **Scalable:** Capable of handling increasing loads as demand grows.
- **Reliable:** Consistent in their performance and able to recover from failures.
- **Responsive:** Able to handle real-time or batch inference requests efficiently.

### Real-World Case Study

**Company:** Uber  
**Use Case:** Ride Pricing  
Uber uses model serving to deploy machine learning models that dynamically price rides based on demand, supply, and other factors. By serving these models through a robust API, Uber ensures that ride pricing remains accurate and responsive to changing conditions.

### Hands-On Code Example

Below is a Python code example using Flask to serve a machine learning model.



**Explanation:**
- **joblib.load('model.pkl')**: Loads the pre-trained machine learning model.
- **@app.route('/predict', methods=['POST'])**: Defines an endpoint for making predictions.
- **request.json**: Retrieves the input data from the HTTP request.
- **model.predict([features])**: Makes a prediction using the loaded model.
- **jsonify({'prediction': prediction.tolist()})****:** Returns the prediction as a JSON response.

```python
from flask import Flask, request, jsonify
import joblib

app = Flask(__name__)

# Load the pre-trained model
model = joblib.load('model.pkl')

@app.route('/predict', methods=['POST'])
def predict():
    data = request.json
    features = data['features']
    prediction = model.predict([features])
    return jsonify({'prediction': prediction.tolist()})

if __name__ == '__main__':
    app.run(host='0.0.0.0', port=5000)
```

## Key Concepts

- **Model Registry:** Centralized repository for storing, versioning, and managing ML models.
- **Model Serving:** Deploying ML models into production for making predictions.
- **Versioning:** Tracking different versions of ML models to enable rollbacks and comparisons.
- **Scalability and Reliability:** Ensuring that served models can handle increasing loads and are robust to failures.
