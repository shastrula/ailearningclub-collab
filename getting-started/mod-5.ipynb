## Running Python Code

Python code can be executed in two primary ways: interactively in the Python shell or by running a script file. Understanding both methods is crucial for effective Python programming.

### Interactive Shell

The interactive shell allows you to execute Python commands one at a time and see the results immediately. This is useful for testing small snippets of code.

### Script Files

Script files (.py) contain a series of Python commands that are executed sequentially when the file is run. This method is ideal for more complex programs.

### Example: Simple Script

Here's a simple Python script to get you started:



**Output:**
```
Hello, AI!
My name is Alice and I am 25 years old
```

````python
# hello.py
# Your first Python program
print('Hello, AI!')

# Variables
name = 'Alice'
age = 25

# Using f-string for formatted output
print(f'My name is {name} and I am {age} years old')
````

## Functions

Functions are reusable blocks of code that perform a specific task. They help in organizing code and making it more readable.

### Defining and Calling Functions

Here’s how you can define and call a function in Python:



**Output:**
```
Hello, Alice!
```

### Functions with Multiple Parameters

Functions can also accept multiple parameters.



**Output:**
```
8
```

````python
# Define a function
def greet(name):
    """
    This function greets a person by name.
    """
    return f'Hello, {name}!'

# Call the function
result = greet('Alice')
print(result)
````

````python
# Function with multiple parameters
def add(a, b):
    """
    This function adds two numbers.
    """
    return a + b

print(add(5, 3))
````

## Real-World Case Study: Data Analysis with Pandas

Pandas is a powerful library for data manipulation and analysis. Let’s look at a simple example of how Pandas is used in a real-world scenario.

### Example: Analyzing Sales Data

Suppose you have a CSV file containing sales data, and you want to analyze it to find the total sales per product.



In this example, Pandas helps in loading, manipulating, and analyzing the sales data efficiently.

````python
import pandas as pd

# Load the dataset
data = pd.read_csv('sales_data.csv')

# Display the first few rows of the dataset
print(data.head())

# Group by product and sum the sales
total_sales = data.groupby('product')['sales'].sum()

print(total_sales)
````
