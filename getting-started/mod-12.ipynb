# Pull Requests and Code Review

## What is a Pull Request?

A **Pull Request (PR)** is a crucial mechanism in collaborative software development that allows developers to propose changes to a repository. It serves as a formal request to merge a set of changes from one branch into another, typically from a feature branch into the main branch. This process facilitates code review, discussion, and approval before changes are integrated into the main codebase.

### Why Use Pull Requests?

- **Code Review:** Enables peers to review code changes, catch bugs, and improve code quality.
- **Collaboration:** Fosters communication and collaboration among team members.
- **Documentation:** Provides a record of changes and discussions, enhancing project transparency.
- **Version Control:** Helps manage different versions and features within the project.

## Creating a Pull Request on GitHub

To create a Pull Request, follow these steps:

### Step 1: Create and Push a Local Branch

First, create a local branch for your feature and push it to GitHub.

```bash title="git-push-branch.sh"
# Create a local branch
git checkout -b feature/add-validation

# Make changes
echo 'def validate_input(x):' >> model.py
echo'   return x is not None' >> model.py

# Commit changes
git add model.py
git commit -m 'Add input validation'

# Push to GitHub
git push origin feature/add-validation
```

### Step 2: Open a Pull Request

On GitHub, you’ll see a prompt to create a PR:

1. Click "Compare & pull request".
2. Add a title: "Add input validation".
3. Add a detailed description explaining your changes.
4. Click "Create pull request".

## Code Review Process

Once your PR is created, team members can engage in the following activities:

- **Review** your code.
- **Comment** on specific lines.
- **Request changes** if needed.
- **Approve** when satisfied.

### Example Review Comment

```
This function looks good, but consider adding error handling for None values.
```

## Responding to Review Comments

To address review comments, make the necessary changes and push them to the same branch.

```bash title="git-update-pr.sh"
# Make the requested changes
echo 'def validate_input(x):' > model.py
echo'   if x is None:' >> model.py
echo'       raise ValueError("Input cannot be None")' >> model.py
echo'   return True' >> model.py

# Commit and push
git add model.py
git commit -m 'Add error handling for None values'
git push origin feature/add-validation
```

The PR automatically updates with your new commits!

## Merging a Pull Request

Once your PR is approved, merge it into the main branch:

1. Click "Merge pull request" on GitHub.
2. Choose a merge strategy (usually "Create a merge commit").
3. Click "Confirm merge".
4. Delete the branch (optional but recommended).

## Best Practices for Pull Requests

### Small Pull Requests

**Why:** Smaller PRs are easier to review and provide faster feedback.

### Clear Title

**Why:** A clear title helps reviewers understand the change at a glance.

### Detailed Description

**Why:** A detailed description provides context and helps reviewers understand the purpose and scope of the changes.

### Link Issues

**Why:** Linking issues tracks what problem the PR solves and provides a reference for future discussions.

### Request Reviewers

**Why:** Requesting specific reviewers ensures that the right people provide feedback, speeding up the review process.

> **💡 Tip:** Keep PRs focused on one feature. If you're adding validation AND preprocessing, make two separate PRs.

## Hands-On Code Example: Implementing a Simple Pull Request Workflow

### Step 1: Set Up the Repository

```bash
# Clone the repository
git clone https://github.com/your-username/your-repo.git
cd your-repo

# Create a new branch
git checkout -b feature/add-logging
```

### Step 2: Make Changes

```python title="model.py"
# Original code
def train_model(data):
    # Training logic here
    pass

# Updated code with logging
import logging

logging.basicConfig(level=logging.INFO)

def train_model(data):
    logging.info("Starting model training")
    # Training logic here
    logging.info("Model training completed")
```

### Step 3: Commit and Push Changes

```bash
# Stage the changes
git add model.py

# Commit the changes
git commit -m "Add logging to model training"

# Push to GitHub
git push origin feature/add-logging
```

### Step 4: Create a Pull Request

1. Go to the repository on GitHub.
2. Click "Compare & pull request".
3. Add a title: "Add logging to model training".
4. Add a description: "This PR adds logging to the model training process to improve visibility and debugging."
5. Click "Create pull request".
