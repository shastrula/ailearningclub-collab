# Committing Your ML Project

## Organizing Your Project

A well-organized project is crucial for collaboration, maintainability, and reproducibility in machine learning projects. Below is a detailed guide on how to structure your project effectively.

### Project Structure

Let's start by creating a structured directory for our machine learning project.

```bash
# Create project structure
mkdir -p my-ai-project/{data,models,notebooks,src}
cd my-ai-project

# Create initial files
echo 'print("ML Model")' > src/train.py
echo 'data/\nmodels/\n__pycache__/\n*.env\n*.pyc' >.gitignore

# Check status
git status
```

**Explanation:**

- `data/`: Directory to store raw and processed data.
- `models/`: Directory to save trained model files.
- `notebooks/`: Directory for Jupyter notebooks used for experimentation and analysis.
- `src/`: Directory for source code, including scripts for training, evaluation, and deployment.
- `.gitignore`: File to specify which files and directories Git should ignore.

### Why This Structure?

- **Modularity**: Separating code, data, and models makes the project easier to navigate and understand.
- **Reproducibility**: Clear structure helps in reproducing results by providing a consistent environment.
- **Collaboration**: Easy for team members to locate and modify specific parts of the project.

## Committing Your Work

Once your project is structured, the next step is to commit your work to a version control system like Git. This ensures that your changes are tracked, and you can collaborate effectively with others.

### Initializing Git

First, initialize a Git repository and make your first commit.

```bash
# Initialize Git repository
git init

# Stage all changes
git add.

# Commit with a descriptive message
git commit -m 'Initial commit: Set up project structure and add training script'

# Push to GitHub (assuming you have a remote repository set up)
git remote add origin https://github.com/yourusername/my-ai-project.git
git push -u origin main
```

**Explanation:**

- `git init`: Initializes a new Git repository.
- `git add .`: Stages all changes for commit.
- `git commit -m 'message'`: Commits the staged changes with a descriptive message.
- `git remote add origin...`: Adds a remote repository URL.
- `git push -u origin main`: Pushes the changes to the remote repository and sets the upstream branch.

## What is `.gitignore`?

The `.gitignore` file tells Git which files and directories to ignore. This is particularly useful for excluding large files, temporary files, and generated outputs that do not need to be versioned.

### Common Entries in `.gitignore`

- `data/`: Ignore raw and processed data files.
- `models/`: Ignore trained model files.
- `__pycache__/`: Ignore Python bytecode files.
- `.env`: Ignore environment variable files.
- `*.pyc`: Ignore compiled Python files.

**Example `.gitignore` File:**

```plaintext
# Data and model files
data/
models/

# Python cache files
__pycache__/
*.pyc

# Environment variables
.env

# Jupyter Notebook checkpoints
.ipynb_checkpoints/
```

### Committing the Training Script


````python
# src/train.py
import torch
import torch.nn as nn
import torch.optim as optim

# Define a simple neural network
class Net(nn.Module):
    def __init__(self):
        super(Net, self).__init__()
        self.fc1 = nn.Linear(10, 5)
        self.fc2 = nn.Linear(5, 1)

    def forward(self, x):
        x = torch.relu(self.fc1(x))
        x = self.fc2(x)
        return x

# Instantiate the network
net = Net()

# Define a loss function and optimizer
criterion = nn.MSELoss()
optimizer = optim.SGD(net.parameters(), lr=0.01)

# Print the model
print(net)
````
