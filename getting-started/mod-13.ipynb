# Collaborating with Teams

## Setting Up Team Collaboration

Effective collaboration with teams is crucial for the success of any AI project. When working with a team, everyone needs access to the same repository and should follow best practices for version control and communication.

## Adding Team Members to GitHub

To add team members to your GitHub repository:

1. Navigate to your repository on GitHub.
2. Click on **Settings** → **Collaborators**.
3. Click on **Add people**.
4. Enter their GitHub username.
5. Choose their permission level: **Pull** (read-only), **Triage** (manage issues and pull requests), **Push** (write access), **Maintain** (full write access and manage repository settings), or **Admin** (full control including managing collaborators).

## Cloning a Repository

Team members can clone the repository to their local machine using Git:

```bash title="git-clone.sh"
# Clone the repository
git clone https://github.com/username/my-ai-project.git

# Navigate to the project directory
cd my-ai-project

# Check the remote repository URL
git remote -v
```

**Output:**
```
origin  https://github.com/username/my-ai-project.git (fetch)
origin  https://github.com/username/my-ai-project.git (push)
```

## Pulling Latest Changes

Before starting work, always pull the latest changes to ensure your local copy is up-to-date:

```bash title="git-pull.sh"
# Update your local repository
git pull origin main

# Or fetch and merge separately
git fetch origin
git merge origin/main
```

## Workflow for Team Collaboration

Follow this workflow to collaborate effectively:

1. **Pull** the latest changes from the main branch.
2. **Create** a feature branch for your changes.
3. **Make** your changes and commit them.
4. **Push** your changes to GitHub.
5. **Create** a Pull Request (PR).
6. **Review** the PR and get approval from team members.
7. **Merge** the PR into the main branch.

```bash title="team-workflow.sh"
# 1. Pull latest changes
git pull origin main

# 2. Create a feature branch
git checkout -b feature/add-logging

# 3. Make changes and commit
echo 'import logging' >> model.py
git add model.py
git commit -m 'Add logging support'

# 4. Push to GitHub
git push origin feature/add-logging

# 5-7. Create PR, get review, merge on GitHub
```

## Handling Merge Conflicts

When two people edit the same file, Git can't automatically merge the changes, resulting in a merge conflict. Here’s how to resolve it:

**Example Conflict:**
```
CONFLICT (content conflict): Merge conflict in model.py
```

### Resolving Conflicts

```bash title="resolve-conflict.sh"
# Open the conflicted file
# You'll see markers like:
# <<<<<<< HEAD
# your changes
# =======
# their changes
# >>>>>>> branch-name

# Edit the file to keep what you want
# Then mark as resolved
git add model.py
git commit -m 'Resolve merge conflict'
git push origin feature/branch-name
```

## Communication Best Practices

Effective communication is key to successful team collaboration. Here are some best practices:

| Practice | Example |
|----------|---------|
| Clear commit messages | "Add data validation" instead of "fix" |
| Descriptive PR titles | "Add preprocessing pipeline" |
| Link related issues | "Fixes #42" in PR description |
| Comment on code | Explain **why**, not just **what** |
| Use branches for features | Never push directly to main |

## Common Team Scenarios

### Scenario 1: Someone Pushed While You Were Working

If your push is rejected because someone else has pushed changes:

```bash
# Pull the latest changes
git pull origin main
# Resolve any conflicts
git push origin feature/branch
```

### Scenario 2: You Need to Update Your Branch

To rebase your branch on the latest main:

```bash
# Fetch the latest changes
git fetch origin
# Rebase your branch
git rebase origin/main
# Push your changes (use --force-with-lease for safety)
git push origin feature/branch --force-with-lease
```

### Scenario 3: Accidentally Committed to Main

If you accidentally committed to the main branch:

```bash
# Create a new branch from the current state
git branch feature/oops

# Reset main to the previous commit
git reset --hard HEAD~1

# Switch to your feature branch
git checkout feature/oops
```

> **💡 Tip:** Always communicate with your team before force-pushing. Use `--force-with-lease` instead of `--force` for safety.
