# Merging and Rebasing

## Introduction

In this module, we will dive deep into the concepts of **merging** and **rebasing** in Git. These are essential techniques for managing branches and integrating changes in a collaborative environment. We'll cover the "why" and "how" behind these concepts, provide real-world case studies, and include hands-on code examples.

## What is Merging?

**Merging** is the process of combining changes from one branch into another. It is the most common way to integrate work from different branches in Git. Merging preserves the history of both branches, creating a merge commit that ties them together.

### Creating and Switching Branches

Before merging, you need to create a branch to work on:

```bash title="git-branch.sh"
# Create a new branch
git branch feature/add-model

# Switch to the new branch
git checkout feature/add-model

# Or create and switch in one command
git checkout -b feature/add-model

# List all branches
git branch -a
```

Output:
```
* feature/add-model
  main
```

### Making Changes on a Branch

```bash title="git-branch-changes.sh"
# Make some changes
echo 'def train_model():' >> model.py
echo'   pass' >> model.py

# Stage and commit
git add model.py
git commit -m 'Add train_model function'

# View the commit
git log --oneline
```

Output:
```
a1b2c3d (HEAD -> feature/add-model) Add train_model function
e4f5g6h (main) Initial commit
```

### Merging Branches

Once your feature is ready, merge it back into `main`:

```bash title="git-merge.sh"
# Switch back to main
git checkout main

# Merge the feature branch
git merge feature/add-model

# View the merged history
git log --oneline
```

Output:
```
a1b2c3d (HEAD -> main) Add train_model function
e4f5g6h Initial commit
```

## Understanding Rebasing

**Rebasing** is an alternative to merging. It replays your commits on top of another branch, creating a linear history. This can make the project history cleaner and easier to follow.

### Example of Rebasing

```bash title="git-rebase.sh"
# Create a new branch
git checkout -b feature/add-preprocessing

# Make changes and commit
echo 'def preprocess_data():' >> model.py
git add model.py
git commit -m 'Add preprocessing function'

# Rebase onto main
git rebase main

# View the linear history
git log --oneline
```

Output:
```
a1b2c3d (HEAD -> feature/add-preprocessing) Add preprocessing function
e4f5g6h (main) Initial commit
```

## Merge vs Rebase

| Aspect         | Merge                              | Rebase                                   |
|----------------|-------------------------------------|------------------------------------------|
| **History**    | Non-linear, shows branches          | Linear, cleaner                          |
| **Safety**     | Safer, preserves history            | Rewrites history                         |
| **Use Case**   | Team collaboration                  | Local cleanup before pushing changes     |

### Real-World Case Study

**Company XYZ** uses Git for their machine learning projects. They have a `main` branch for production-ready code and feature branches for new developments. 

- **Merging**: When a feature is completed and tested, the team merges it into the `main` branch. This preserves the history of the feature branch and allows them to track changes easily.
- **Rebasing**: Before pushing changes to the remote repository, developers rebase their feature branches onto the latest `main` branch. This ensures a clean, linear history and avoids complex merge commits.

## Hands-On Code Examples

### Merging Example in Python

```python title="merge_example.py"
# Suppose we have a main branch with an initial model
def initial_model():
    print("This is the initial model.")

# Feature branch adds a new function
def train_model():
    print("Training the model.")

# Merging the feature branch into main
# In a real scenario, this would be done via Git commands as shown above
# Here, we simulate the merge by simply calling both functions
initial_model()
train_model()
```

### Rebasing Example in Python

```python title="rebase_example.py"
# Main branch with an initial model
def initial_model():
    print("This is the initial model.")

# Feature branch adds a preprocessing function
def preprocess_data():
    print("Preprocessing data.")

# Rebasing the feature branch onto main
# In a real scenario, this would be done via Git commands as shown above
# Here, we simulate the rebase by simply calling both functions
initial_model()
preprocess_data()
```
