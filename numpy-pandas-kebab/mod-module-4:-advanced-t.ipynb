## Key Concept: Optimization Techniques

Optimization techniques play a crucial role in data manipulation. We will explore different methods to optimize the performance of our code, such as using caching, parallel processing, and vectorization.

```python title="optimize.py"
# Optimized code using NumPy's vectorized operations
import numpy as np
def calculate_mean(numbers):
    return np.mean(numbers)
numbers = [1, 2, 3, 4, 5]
calculate_mean(numbers)
```

> **Try it in Google Colab:** [![Open in Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/shastrula/ailearningclub-courses/blob/main/numpy-pandas-kebab/mod-module-4:-advanced-t.ipynb)


> **💡 Tip:** When working with large datasets, it's essential to use optimized techniques to avoid slow performance. Caching can significantly improve the execution time of your code.

<div class="quiz" data-correct="1">
  <p class="font-semibold mb-3">❓ What is the purpose of caching in optimization?</p>
  <div class="space-y-2">
    <label class="flex items-center gap-2 cursor-pointer">
      <input type="radio" name="q4372652352" value="0">
      <span>To reduce data size</span>
    </label>
    <label class="flex items-center gap-2 cursor-pointer">
      <input type="radio" name="q4372652352" value="1">
      <span>To speed up computation</span>
    </label>
    <label class="flex items-center gap-2 cursor-pointer">
      <input type="radio" name="q4372652352" value="2">
      <span>To increase memory usage</span>
    </label>
  </div>
  <button class="quiz-btn mt-3 px-4 py-2 bg-blue-600 text-white rounded text-sm font-medium hover:bg-blue-700">Check Answer</button>
  <p class="quiz-result text-sm mt-2 hidden"></p>
</div>

<div class="quiz" data-correct="3">
  <p class="font-semibold mb-3">❓ Which type of optimization technique is suitable for datasets with missing values?</p>
  <div class="space-y-2">
    <label class="flex items-center gap-2 cursor-pointer">
      <input type="radio" name="q4372651840" value="0">
      <span>Vectorization</span>
    </label>
    <label class="flex items-center gap-2 cursor-pointer">
      <input type="radio" name="q4372651840" value="1">
      <span>Caching</span>
    </label>
    <label class="flex items-center gap-2 cursor-pointer">
      <input type="radio" name="q4372651840" value="2">
      <span>Parallel processing</span>
    </label>
    <label class="flex items-center gap-2 cursor-pointer">
      <input type="radio" name="q4372651840" value="3">
      <span>Data Imputation</span>
    </label>
  </div>
  <button class="quiz-btn mt-3 px-4 py-2 bg-blue-600 text-white rounded text-sm font-medium hover:bg-blue-700">Check Answer</button>
  <p class="quiz-result text-sm mt-2 hidden"></p>
</div>
