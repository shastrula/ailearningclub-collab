# Module 2: Data Structures and Operations in Numpy

## Key Concept

NumPy arrays: one-dimensional, multi-dimensional, and sparse arrays; array indexing, slicing, and reshaping.

```python title="example.py"
# Import the NumPy library
import numpy as np
# Create a 1D array with numbers from 0 to 9
array_1d = np.arange(10)
# Create a 2D array with numbers from 0 to 9
array_2d = np.arange(4).reshape((2, 2))
```

> **Try it in Google Colab:** [![Open in Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/shastrula/ailearningclub-courses/blob/main/numpy-pandas-kebab/mod-module-2:-data-struc.ipynb)


> **💡 Tip:** When working with large datasets, NumPy arrays can significantly speed up your code and improve performance.

<div class="quiz" data-correct="0">
  <p class="font-semibold mb-3">❓ What does the `np.newaxis` parameter do?</p>
  <div class="space-y-2">
    <label class="flex items-center gap-2 cursor-pointer">
      <input type="radio" name="q4372645184" value="0">
      <span>Adds an extra dimension to the array.</span>
    </label>
    <label class="flex items-center gap-2 cursor-pointer">
      <input type="radio" name="q4372645184" value="1">
      <span>Removes a dimension from the array.</span>
    </label>
    <label class="flex items-center gap-2 cursor-pointer">
      <input type="radio" name="q4372645184" value="2">
      <span>Creates a copy of the original array.</span>
    </label>
    <label class="flex items-center gap-2 cursor-pointer">
      <input type="radio" name="q4372645184" value="3">
      <span>Resizes the array to fit in memory.</span>
    </label>
  </div>
  <button class="quiz-btn mt-3 px-4 py-2 bg-blue-600 text-white rounded text-sm font-medium hover:bg-blue-700">Check Answer</button>
  <p class="quiz-result text-sm mt-2 hidden"></p>
</div>

<div class="quiz" data-correct="1">
  <p class="font-semibold mb-3">❓ Which operation is used to multiply two NumPy arrays?</p>
  <div class="space-y-2">
    <label class="flex items-center gap-2 cursor-pointer">
      <input type="radio" name="q4372659712" value="0">
      <span>Matrix multiplication</span>
    </label>
    <label class="flex items-center gap-2 cursor-pointer">
      <input type="radio" name="q4372659712" value="1">
      <span>Element-wise multiplication</span>
    </label>
    <label class="flex items-center gap-2 cursor-pointer">
      <input type="radio" name="q4372659712" value="2">
      <span>Array indexing</span>
    </label>
    <label class="flex items-center gap-2 cursor-pointer">
      <input type="radio" name="q4372659712" value="3">
      <span>Array slicing</span>
    </label>
  </div>
  <button class="quiz-btn mt-3 px-4 py-2 bg-blue-600 text-white rounded text-sm font-medium hover:bg-blue-700">Check Answer</button>
  <p class="quiz-result text-sm mt-2 hidden"></p>
</div>
