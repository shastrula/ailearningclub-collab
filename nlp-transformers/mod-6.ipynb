# Basics of Tokenization

## Understanding Tokenization

Tokenization is the process of converting a sequence of text into tokens. These tokens can be words, subwords, or characters, depending on the tokenizer used. The choice of tokenizer can significantly impact the performance of NLP models, as it affects how the model interprets and processes the input text. 

### Why Tokenization Matters

Tokenization is critical because it transforms raw text data into a structured format that machine learning models can process. By breaking down text into smaller, manageable pieces, tokenization helps models understand the context and meaning of words within sentences. This is especially important for tasks like sentiment analysis, machine translation, and text classification.

### How Tokenization Works

Tokenization can be performed at different levels:
- **Word Tokenization:** Splitting text into individual words.
- **Subword Tokenization:** Splitting text into subwords, which are smaller units than words. This is useful for handling out-of-vocabulary (OOV) words.
- **Character Tokenization:** Splitting text into individual characters.

## Tokenization in Practice

### Word Tokenization

Word tokenization is the simplest form of tokenization, where text is split into individual words. 



**Output:**
```
['Tokenization', 'is', 'essential', 'in', 'NLP', '.']
```

### Subword Tokenization

Subword tokenization is more sophisticated and is used by modern NLP models like BERT. It splits words into subwords, which helps handle OOV words more effectively.



**Output:**
```
['token', 'ization', 'is', 'essen', 'tial', 'in', 'nlp', '.']
```

````python
import nltk
nltk.download('punkt')
from nltk.tokenize import word_tokenize

# Sample text
sample_text = "Tokenization is essential in NLP."

# Tokenize the text into words
tokens = word_tokenize(sample_text)

print(tokens)
````

````python
from transformers import BertTokenizer

# Initialize the BERT tokenizer
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

# Sample text
sample_text = "Tokenization is essential in NLP."

# Tokenize the text into subwords
tokens = tokenizer.tokenize(sample_text)

print(tokens)
````

## Tokenization with Special Tokens

Special tokens like `[CLS]`, `[SEP]`, and `` are used in tokenization to provide additional context or to handle specific tasks. 

- **[CLS]**: Used to aggregate information for classification tasks.
- **[SEP]**: Used to separate different sentences or inputs.
- **[MASK]**: Used for masked language modeling tasks.

### Example with Special Tokens



**Output:**
```
[101, 7901, 2003, 2000, 1996, 102, 101, 7901, 2024, 2000, 1996, 102]
```

````python
from transformers import BertTokenizer

# Initialize the BERT tokenizer
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

# Sample text
sample_text = "Tokenization is essential in NLP. BERT uses special tokens."

# Tokenize the text with special tokens
tokens = tokenizer.tokenize(sample_text)
tokens = ['[CLS]'] + tokens + ['[SEP]']

# Convert tokens to input IDs
input_ids = tokenizer.convert_tokens_to_ids(tokens)

print(input_ids)
````
