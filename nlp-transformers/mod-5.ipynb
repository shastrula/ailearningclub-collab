# Setting Up the Environment

## Installing Python and Virtual Environment

Before diving into NLP, ensure Python is installed on your system. We recommend using a virtual environment to manage dependencies. This keeps your global Python installation clean and avoids conflicts between project requirements.

### Why Use a Virtual Environment?

A virtual environment creates an isolated environment for Python projects. This means that each project can have its own dependencies, regardless of what dependencies every other project has.

### How to Set Up a Virtual Environment

Here's a step-by-step guide to creating and activating a virtual environment:



### Real-World Case Study

At **Company X**, developers use virtual environments to manage different projects that require different versions of libraries. This practice has significantly reduced the time spent on resolving dependency conflicts.

```python
import os
import subprocess

# Create a virtual environment named 'nlp-env'
os.system('python -m venv nlp-env')

# Activate the virtual environment
# On Windows:
# nlp-env\Scripts\activate
# On Unix or MacOS:
subprocess.run(['bash', '-c','source nlp-env/bin/activate'])

# Upgrade pip to the latest version
subprocess.run(['pip', 'install', '--upgrade', 'pip'])
```

## Installing Required Libraries

Next, install the essential libraries for NLP and Transformers. We'll use `transformers` from HuggingFace and `torch` for PyTorch support.

### Why These Libraries?

- **Transformers**: Provides easy access to pre-trained models like BERT, making it simpler to implement state-of-the-art NLP techniques.
- **PyTorch**: A powerful library for deep learning that integrates well with HuggingFace's Transformers.

### How to Install Them



### Real-World Case Study

At **Company Y**, the NLP team uses HuggingFace's Transformers library to quickly prototype and deploy BERT-based models for sentiment analysis in customer reviews.

```python
import subprocess

# Install transformers and torch
subprocess.run(['pip', 'install', 'transformers', 'torch'])

# Verify installation
import transformers
import torch

print(f'Transformers version: {transformers.__version__}')
print(f'Torch version: {torch.__version__}')
```

## Hands-On Code Example

Let's put it all together in a script that sets up the environment and verifies the installation.

```python
import os
import subprocess

# Create and activate virtual environment
os.system('python -m venv nlp-env')
subprocess.run(['bash', '-c','source nlp-env/bin/activate'])
subprocess.run(['pip', 'install', '--upgrade', 'pip'])

# Install required libraries
subprocess.run(['pip', 'install', 'transformers', 'torch'])

# Verify installation
import transformers
import torch

print(f'Transformers version: {transformers.__version__}')
print(f'Torch version: {torch.__version__}')
```
