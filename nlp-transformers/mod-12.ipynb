# Deploying Transformer Models

## Loading and Using Pre-trained BERT Models

### Why Use Pre-trained Models?

Pre-trained BERT models are powerful tools for a variety of NLP tasks. They are trained on large datasets and can be fine-tuned for specific applications. This saves time and computational resources compared to training a model from scratch. Using HuggingFace's Transformers library, we can easily load and utilize these models.

### How to Load and Use BERT

Here's a step-by-step guide to loading and using a pre-trained BERT model.



**Explanation:**
- **Tokenizer:** Converts text into tokens that the model understands.
- **Model:** Pre-trained BERT model loaded from HuggingFace.
- **Encoding:** Converts input text into tensors.
- **Outputs:** Feeds the encoded text into the model to get embeddings.

```python
from transformers import BertTokenizer, BertModel
import torch

# Load pre-trained BERT model and tokenizer
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')
model = BertModel.from_pretrained('bert-base-uncased')

# Encode a sample text
inputs = tokenizer("Hello, how are you?", return_tensors='pt')

# Get the embeddings
outputs = model(**inputs)

# Print the embeddings
print(outputs.last_hidden_state)
```

## Fine-tuning a Pre-trained BERT Model

### Why Fine-tune?

Fine-tuning a pre-trained BERT model involves training it on a specific dataset to adapt it to a particular task, such as sentiment analysis or named entity recognition. This process allows the model to leverage its pre-trained knowledge while adapting to the nuances of the specific task.

### How to Fine-tune BERT

Here's a step-by-step guide to fine-tuning a pre-trained BERT model.



**Explanation:**
- **Dataset:** Loads the GLUE MRPC dataset.
- **Model:** Loads a pre-trained BERT model for sequence classification.
- **Training Arguments:** Defines hyperparameters for training.
- **Trainer:** Initializes the Trainer object with the model and training arguments.
- **Training:** Starts the training process.

```python
from transformers import BertForSequenceClassification, Trainer, TrainingArguments
from datasets import load_dataset

# Load a dataset
dataset = load_dataset('glue','mrpc')

# Load pre-trained BERT model
model = BertForSequenceClassification.from_pretrained('bert-base-uncased')

# Define training arguments
training_args = TrainingArguments(
    output_dir='./results',
    evaluation_strategy='epoch',
    learning_rate=2e-5,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=64,
    num_train_epochs=3,
    weight_decay=0.01,
)

# Initialize Trainer
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=dataset['train'],
    eval_dataset=dataset['validation']
)

# Train the model
trainer.train()
```

## Real-World Case Study: Sentiment Analysis in E-commerce
