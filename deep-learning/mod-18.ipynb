# Working with GPUs

## Introduction

Utilizing GPUs for deep learning can significantly accelerate model training, making it an essential skill for efficient and scalable projects. This module will delve into the essentials of working with GPUs in PyTorch, including checking GPU availability, moving data to the GPU, and practical case studies.

## Checking GPU Availability

Before utilizing a GPU, it's crucial to verify its availability. PyTorch offers a straightforward method to check for GPU availability using the `torch.cuda.is_available()` function. This function returns a boolean indicating whether CUDA—NVIDIA's parallel computing platform—is available.

### Why Check GPU Availability?

- **Resource Management:** Ensures your code runs on the appropriate device, avoiding unnecessary errors.
- **Conditional Execution:** Allows you to write code that can run on both CPU and GPU based on availability.

### How to Check GPU Availability

```python
import torch

# Check if CUDA (GPU support) is available
is_gpu_available = torch.cuda.is_available()

print(f'Is CUDA available: {is_gpu_available}')

# If CUDA is available, print the number of GPUs
if is_gpu_available:
    print(f'Number of GPUs: {torch.cuda.device_count()}')
```

## Moving Data to GPU

Once you've confirmed that a GPU is available, the next step is to transfer your data and models to the GPU. PyTorch allows you to move tensors and models to the GPU using the `.to(device)` method, where `device` is a `torch.device` object specifying the GPU.

### Why Move Data to GPU?

- **Performance:** GPUs are optimized for parallel computations, making them much faster for training deep learning models.
- **Efficiency:** Utilizing GPUs can reduce training time from hours to minutes.

### How to Move Data to GPU



**💡 Tip:** Always ensure that both your model and your data are on the same device to avoid errors during computation.

```python
import torch

# Check if CUDA is available
is_gpu_available = torch.cuda.is_available()
device = torch.device('cuda' if is_gpu_available else 'cpu')

# Create a tensor and move it to the GPU
tensor_cpu = torch.tensor([1.0, 2.0, 3.0])
tensor_gpu = tensor_cpu.to(device)

print(f'Tensor on {device}:\n{tensor_gpu}')
```
