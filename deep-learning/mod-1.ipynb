# Introduction to Deep Learning

## 1. What is Deep Learning?

Deep learning is a specific approach to machine learning based on **Artificial Neural Networks (ANNs)**. The "deep" in deep learning refers to the number of layers in these networks. 

### Why Deep Learning?

While traditional machine learning algorithms (like Linear Regression or Random Forests) plateau in performance as you feed them more data, deep learning models thrive on massive datasets. They can automatically discover representations and features from raw data, eliminating the need for manual feature engineering in many complex tasks like image and speech recognition.

### Real-World Case Study: Self-Driving Cars

Self-driving cars rely on deep learning models to interpret sensor data (cameras, LiDAR, radar) and make real-time decisions. Companies like Tesla use deep learning to detect objects, recognize traffic signs, and navigate roads, showcasing the power and necessity of these models in critical applications.

## 2. Anatomy of a Neural Network

A neural network is composed of interconnected nodes (neurons) organized into layers:

```text
Input Layer    Hidden Layers    Output Layer
    ●              ●                ●
    ●              ●                ●
    ●              ●                ●
    ●              ●                ●
    
    x₁            h₁,h₂,h₃          y
    x₂            (neurons)      (prediction)
    x₃
    x₄
    
Forward Pass: x → h → y
Backward Pass: ∂L/∂w ← gradients
```

### Key Components:

- **Input Layer:** Receives the raw data (e.g., pixels of an image, words in a sentence).
- **Hidden Layers:** The layers between input and output where the "magic" happens. They extract patterns from the data. A network is considered "deep" if it has more than one hidden layer.
- **Neurons (Nodes):** The basic computational unit. Each neuron receives inputs, multiplies them by learned **weights**, adds a **bias**, and passes the result through an **activation function**.
- **Weights and Biases:** The learnable parameters of the network. Training a network means adjusting these values to minimize errors.
- **Activation Function:** Introduces non-linearity. Without this, no matter how many layers you have, the network would just be a simple linear regression model. Common activation functions include ReLU (Rectified Linear Unit), Sigmoid, and Tanh.

## 3. The Learning Process: Forward and Backward Passes

Training a neural network is an iterative process of making predictions, measuring errors, and adjusting weights.

### Forward Pass

Data flows from the input layer, through the hidden layers, to the output layer to make a prediction.

### Loss Calculation

A **Loss Function** measures the difference between the network's prediction and the actual true value (the ground truth).

### Backward Pass (Backpropagation)

The network calculates the gradient (derivative) of the loss function with respect to every weight in the network. This tells the network how to adjust each weight to reduce the error.

### Optimization

An **Optimizer** (like Stochastic Gradient Descent or Adam) updates the weights using the gradients calculated during backpropagation.

### Real-World Case Study: Voice Recognition

Voice assistants like Amazon's Alexa use deep learning models to convert speech to text and understand user commands. These models are trained using large datasets of voice recordings, and they rely on the forward and backward passes to continuously improve their accuracy.

## 4. Hands-On: Your First PyTorch Neural Network

Let's build a simple neural network using PyTorch. This network will take an input vector of 10 numbers and predict a single output value.

```python title="pytorch_basics.py"
import torch
import torch.nn as nn
import torch.optim as optim

# 1. Define the Neural Network Architecture
class SimpleNN(nn.Module):
    def __init__(self):
        super(SimpleNN, self).__init__()
        # First fully connected (linear) layer: 10 inputs -> 5 outputs
        self.fc1 = nn.Linear(10, 5)
        # Second fully connected layer: 5 inputs -> 1 output
        self.fc2 = nn.Linear(5, 1)

    def forward(self, x):
        # Apply ReLU activation after the first layer
        x = torch.relu(self.fc1(x))
        # No activation on the final layer (typical for regression tasks)
        x = self.fc2(x)
        return x

# 2. Instantiate the model
model = SimpleNN()
print("Model Architecture:\n", model)

# 3. Define Loss Function and Optimizer
# Mean Squared Error (MSE) is common for predicting continuous values
criterion = nn.MSELoss()
# Stochastic Gradient Descent (SGD) will update the weights
optimizer = optim.SGD(model.parameters(), lr=0.01)

# 4. Create dummy data for training
# A batch of 1 sample, with 10 features
input_data = torch.randn(1, 10)
# The true value we want the model to predict
target = torch.tensor([[0.5]])

# --- ONE TRAINING STEP ---

# Step A: Forward pass (make a prediction)
prediction = model(input_data)
print(f"\nInitial Prediction: {prediction.item():.4f}")

# Step B: Calculate the loss (error)
loss = criterion(prediction, target)
print(f"Initial Loss: {loss.item():.4f}")

# Step C: Backward pass (calculate gradients)
optimizer.zero_grad() # Clear old gradients
loss.backward()       # Compute new gradients

# Step D: Update weights
optimizer.step()      # Adjust weights based on gradients

# Let's see if the prediction improved after one step!
new_prediction = model(input_data)
new_loss = criterion(new_prediction, target)
print(f"\nPrediction after 1 training step: {new_prediction.item():.4f}")
print(f"Loss after 1 training step: {new_loss.item():.4f}")
```

> **Try it in Google Colab:** [![Open in Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/shastrula/ailearningclub-courses/blob/main/deep-learning/mod-1.ipynb)

**What's happening here?** 
In a real scenario, you would put the training step (Forward -> Loss -> Backward -> Step) inside a `for` loop, repeating it thousands of times across an entire dataset until the loss becomes very small.

## Practice Quizzes

Test your understanding of the core concepts.

### Quiz 1: What is the primary function of an activation function in a neural network?
- [ ] To compute the weighted sum of inputs and scale them to zero.
- [✓] To introduce non-linearity, allowing the network to learn complex, non-linear patterns.
- [ ] To update the weights during the backpropagation step.
- [ ] To normalize the input data before it enters the first layer.

### Quiz 2: During the training process, what is the specific role of Backpropagation?
- [ ] It calculates the final prediction of the neural network.
- [✓] It computes the gradients of the loss function with respect to the network's weights, indicating how they should be updated.
- [ ] It randomly initializes the weights and biases of the hidden layers.
- [ ] It measures the total error between the prediction and the ground truth.

### Quiz 3: Why do Deep Learning models often outperform traditional Machine Learning models on complex tasks like image recognition?
- [ ] Deep learning models require significantly less data to train.
- [✓] Deep learning models can automatically discover and learn complex features from raw data, bypassing manual feature engineering.
- [ ] Deep learning models only use linear equations, making them mathematically simpler.
- [ ] Deep learning models do not require an optimization algorithm.
