# Tensors and Autograd

## Tensors

### What are Tensors?

Tensors are the primary data structure used in PyTorch for all computations. They are similar to NumPy's ndarrays but come with the added capability of leveraging GPUs for accelerated computations. Tensors can be scalars, vectors, matrices, or higher-dimensional arrays.

### Why Use Tensors?

Tensors are essential because they allow for efficient mathematical operations that are fundamental to neural network computations. They support a multitude of operations, including element-wise operations, matrix multiplications, and more.

### Creating Tensors

Tensors can be created in various ways, including from data, random numbers, or existing data structures like NumPy arrays.



### Real-World Case Study: Image Processing

In image processing, tensors are used to represent images. Each pixel in an image is a value in a tensor. For example, a grayscale image can be represented as a 2D tensor, while a color image can be represented as a 3D tensor with dimensions (height, width, channels).

```python
import torch
import numpy as np

# Creating a tensor from a list
tensor_from_list = torch.tensor([1, 2, 3])
print('Tensor from list:', tensor_from_list)

# Creating a tensor with random numbers
random_tensor = torch.rand(3, 3)
print('Random tensor:', random_tensor)

# Creating a tensor from a NumPy array
numpy_array = np.array([1, 2, 3])
tensor_from_numpy = torch.from_numpy(numpy_array)
print('Tensor from NumPy array:', tensor_from_numpy)
```

## Autograd

### What is Autograd?

Autograd is PyTorch's automatic differentiation library that powers neural network training. It records operations performed on tensors and allows for the computation of gradients. This is essential for backpropagation, the method by which neural networks learn.

### How Does Autograd Work?

Autograd works by building a computational graph on-the-fly. In this graph, nodes represent tensors, and edges represent functions. When you call `.backward()` on a tensor, Autograd traverses this graph to compute the gradients.

### Key Concepts

- **`requires_grad`**: A tensor attribute that indicates whether gradient tracking is enabled for that tensor.
- **Computational Graph**: A directed graph representing the operations performed on tensors.
- **`.backward()`**: A method that performs backpropagation and computes the gradients.
- **`.grad`**: An attribute that stores the computed gradient for a tensor.
- **`.detach()`**: A method that returns a new tensor detached from the current computation graph, used primarily during inference.

### Example: Computing Gradients



### Real-World Case Study: Training a Neural Network

In training a neural network, Autograd is used to compute the gradients of the loss function with respect to the model parameters. These gradients are then used to update the parameters in the direction that minimizes the loss.

```python
import torch

# Create tensors with requires_grad=True
x = torch.tensor([1.0, 2.0, 3.0], requires_grad=True)
y = torch.tensor([4.0, 5.0, 6.0])

# Perform operations
z = x + y
out = z * z

# Compute gradients
out.backward(torch.tensor([1.0, 1.0, 1.0]))

# Print gradients
print('Gradient of x:', x.grad)
```
