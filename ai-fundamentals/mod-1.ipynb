# Introduction to AI and Machine Learning

## 1. Demystifying AI, ML, and Deep Learning

### Artificial Intelligence (AI)

**Definition:** AI is the broad concept of creating machines that can perform tasks requiring human-like intelligence. This includes problem-solving, understanding natural language, recognizing patterns, and making decisions.

**In-depth Explanation:** AI encompasses a wide range of technologies and methodologies aimed at simulating human intelligence. It involves developing algorithms and models that can reason, learn, and adapt to new inputs. AI systems can be rule-based, where explicit instructions are programmed, or they can be data-driven, where the system learns from examples. 

**Real-World Case Study:** One prominent example of AI is the use of autonomous vehicles. Companies like Tesla employ AI to enable their cars to navigate and make decisions on the road, such as recognizing traffic signs, detecting obstacles, and making real-time driving decisions. 

**Example:** A chess-playing AI evaluates all possible moves using predefined rules and strategies. It uses algorithms like the Minimax algorithm to determine the best move by considering all possible future scenarios.

### Machine Learning (ML)

**Definition:** ML is a subset of AI where algorithms learn patterns from data without being explicitly programmed.

**In-depth Explanation:** Machine Learning involves training models on data so that they can make predictions or decisions without being explicitly instructed to perform the task. This learning process involves feeding the model large amounts of data, allowing it to identify patterns and relationships. Once trained, the model can make predictions or decisions on new, unseen data.

**Real-World Case Study:** Recommendation systems used by streaming services like Netflix are a practical application of ML. These systems analyze user behavior, such as viewing history and ratings, to suggest content that the user is likely to enjoy.

**How it works:** 
1. **Data Input:** The model is fed large amounts of data.
2. **Learning Process:** The algorithm identifies patterns and relationships within the data.
3. **Prediction/Decision Making:** The model uses learned patterns to make predictions or decisions on new data.

**Example:** A recommendation system on a streaming service learns your viewing preferences by analyzing your past behavior and suggesting similar content.

### Deep Learning (DL)

**Definition:** DL is a subset of ML that uses neural networks with multiple layers to model complex relationships in data.

**In-depth Explanation:** Deep Learning involves training artificial neural networks that have multiple layers (hence "deep"). Each layer transforms the input data into increasingly abstract representations, allowing the network to learn complex patterns and relationships. DL has been particularly successful in tasks involving large amounts of unstructured data, such as images, audio, and text.

**Real-World Case Study:** Facial recognition systems used by social media platforms like Facebook employ DL to identify and verify individuals from images or video streams. These systems can recognize faces in photos, tag users, and even unlock devices using facial biometrics.

**How it works:**
1. **Neural Network Structure:** Comprises an input layer, multiple hidden layers, and an output layer.
2. **Learning Process:** Each layer extracts increasingly abstract representations of the data.
3. **Applications:** Powers advanced applications like image and speech recognition, natural language processing, and autonomous driving.

**Example:** Facial recognition systems use DL to identify and verify individuals from images or video streams.

## 2. The Machine Learning Workflow

### Step-by-Step Process

1. **Data Collection:**
   - **Why:** Gather relevant data from various sources (databases, APIs, sensors).
   - **Example:** Collecting customer purchase history from an e-commerce platform.

2. **Data Preprocessing:**
   - **Why:** Clean and prepare the data for analysis.
   - **Steps:**
     - Handle missing values.
     - Remove outliers.
     - Normalize or scale features.
   - **Example:** Filling missing values in a dataset with the mean of the column.

3. **Feature Engineering:**
   - **Why:** Create new features or transform existing ones to improve model performance.
   - **Example:** Converting a date into separate features like `day_of_week` and `month`.

4. **Model Training:**
   - **Why:** Feed preprocessed data into an algorithm to learn patterns.
   - **Example:** Using a decision tree algorithm to predict customer churn.

5. **Model Evaluation:**
   - **Why:** Assess the model’s performance on unseen data.
   - **Metrics:** Accuracy, precision, recall, F1 score.
   - **Example:** Testing a classification model using a held-out test set.

6. **Model Deployment:**
   - **Why:** Integrate the model into a production environment for real-time predictions.
   - **Example:** Deploying a fraud detection model to a banking application.

### Real-World Case Study: Predicting House Prices

**Scenario:** You are tasked with predicting house prices for a real estate company.

1. **Data Collection:**
   - Gather historical sales data including price, square footage, number of bedrooms, and zip code.

2. **Data Preprocessing:**
   - Remove rows with missing prices.
   - Correct typos in zip codes.

3. **Feature Engineering:**
   - Calculate `price_per_sqft` as a new feature.

4. **Model Training:**
   - Use a Linear Regression algorithm to learn the relationship between features and house prices.

5. **Model Evaluation:**
   - Test the model on new, unseen data to ensure predictions are accurate within a 5% margin of error.

6. **Model Deployment:**
   - Deploy the model via a web server API, allowing real estate agents to input house details and receive instant price suggestions.

**Objective:** Predict salary based on years of experience.

### Feature Engineering Example

**Objective:** Scale features to improve model performance.



**Output:**
```
Original Data:
    age   salary
0   25    50000
1   32    65000
2   47   120000
3   51   135000
4   22    48000

Scaled Data (Mean=0, Variance=1):
[[-1.34164079 -1.20902936]
 [-0.53665631 -0.38167939]
 [ 1.07331263  1.44670874]
 [ 1.34164079  1.73205081]
 [-1.47227037 -1.52504071]]
```

**Pro Tip:** Always fit your scaler on your training data and then transform both training and test data to avoid data leakage.

````python
import numpy as np

# Raw Data
x = np.array([1, 2, 3, 4, 5])  # years of experience
y = np.array([40, 50, 60, 70, 80])  # salary in thousands

# Training the Model
# Calculate slope (m) and intercept (b) for the line of best fit
slope = (np.mean(x * y) - np.mean(x) * np.mean(y)) / (np.mean(x**2) - np.mean(x)**2)
intercept = np.mean(y) - slope * np.mean(x)

print(f'Learned Slope (m): {slope}')
print(f'Learned Intercept (b): {intercept}')

# Prediction
new_x = 6  # years of experience
predicted_y = slope * new_x + intercept
print(f'Predicted salary for 6 years experience: ${predicted_y}k')
````

````python
import pandas as pd
from sklearn.preprocessing import StandardScaler

# Raw data
data = {
    'age': [25, 32, 47, 51, 22],
   'salary': [50000, 65000, 120000, 135000, 48000]
}
df = pd.DataFrame(data)

# Standardizing features
scaler = StandardScaler()
scaled_features = scaler.fit_transform(df)

print("Original Data:\n", df)
print("\nScaled Data (Mean=0, Variance=1):\n", scaled_features)
````
