# Unsupervised Learning: Clustering

## K-Means Clustering

### In-Depth Explanation

**K-Means** is a popular clustering algorithm that partitions data into K distinct, non-overlapping subsets. Each subset represents a cluster, and the algorithm aims to minimize the variance within each cluster. This is achieved through an iterative process:

1. **Initialization**: Randomly initialize K centroids.
2. **Assignment**: Assign each data point to the nearest centroid.
3. **Update**: Recalculate the centroids based on the assigned points.
4. **Iteration**: Repeat the assignment and update steps until the centroids stabilize.

### Why K-Means?

K-Means is widely used because of its simplicity and efficiency. It is particularly effective when clusters are compact and spherical, and the number of clusters is known beforehand. However, it may not perform well with irregularly shaped clusters or when the number of clusters is unknown.

### Real-World Case Study

**Customer Segmentation**: E-commerce companies use K-Means to segment customers based on purchasing behavior. This helps in targeted marketing and personalized recommendations. For instance, Amazon uses K-Means to group customers into segments like "frequent buyers," "occasional buyers," and "deal hunters," allowing for tailored promotional strategies.

### Hands-On Code Example



**Output:**
```
Cluster labels: [0 0 0 1 1 1]
Centroids: [[1. 1.33333333]
 [4. 2.        ]]
```

````python
from sklearn.cluster import KMeans
import numpy as np
import matplotlib.pyplot as plt

# Generate sample data
X = np.array([[1, 2], [1, 4], [1, 0],
              [4, 2], [4, 4], [4, 0]])

# Apply K-Means clustering
kmeans = KMeans(n_clusters=2, random_state=0)
kmeans.fit(X)

# Print cluster labels and centroids
print('Cluster labels:', kmeans.labels_)
print('Centroids:', kmeans.cluster_centers_)

# Plot the clusters and centroids
plt.scatter(X[:, 0], X[:, 1], c=kmeans.labels_, cmap='viridis')
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], s=300, c='red')
plt.title('K-Means Clustering')
plt.show()
````

## Hierarchical Clustering

### In-Depth Explanation

**Hierarchical clustering** creates a tree of clusters, which can be visualized using a dendrogram. It does not require specifying the number of clusters beforehand. There are two main types:

1. **Agglomerative (Bottom-Up)**: Starts with each data point as a separate cluster and merges them into larger clusters iteratively.
2. **Divisive (Top-Down)**: Starts with all data points in one cluster and splits them into smaller clusters iteratively.

### Why Hierarchical Clustering?

Hierarchical clustering is useful when the hierarchical structure of the data is important. It allows for visualization of the clustering process and does not require pre-specification of the number of clusters. This makes it particularly useful in scenarios where the relationships between clusters are as important as the clusters themselves.

### Real-World Case Study

**Document Clustering**: News agencies use hierarchical clustering to group similar news articles, aiding in content organization and recommendation systems. For example, BBC uses hierarchical clustering to categorize news articles into topics like "Politics," "Sports," and "Entertainment," which helps in creating a more organized and navigable news portal.

### Hands-On Code Example

````python
from sklearn.cluster import AgglomerativeClustering
import numpy as np
from scipy.cluster.hierarchy import dendrogram, linkage
import matplotlib.pyplot as plt

# Generate sample data
X = np.array([[1, 2], [1, 4], [1, 0],
              [4, 2], [4, 4], [4, 0]])

# Apply Agglomerative Clustering
clustering = AgglomerativeClustering(n_clusters=2)
clustering.fit(X)

# Plot dendrogram
linked = linkage(X, 'ward')
plt.figure(figsize=(10, 7))
dendrogram(linked,
           orientation='top',
           distance_sort='descending',
           show_leaf_counts=True)
plt.title('Hierarchical Clustering Dendrogram')
plt.show()
````
