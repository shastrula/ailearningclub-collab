## Gradient Boosting Machines (GBM)

### In-depth Explanation

**Gradient Boosting Machines (GBM)** are a powerful ensemble technique that combines multiple weak learners to create a strong learner. The core idea behind GBM is to fit new models to the residual errors of preceding models, thereby improving the overall predictive performance iteratively. 

### Why GBM?

- **Iterative Improvement**: Each new model focuses on the errors made by the previous models, leading to gradual and cumulative improvement in performance.
- **Flexibility**: GBM can handle various types of loss functions, making it versatile for different types of problems (regression, classification, ranking).
- **Robustness**: By combining multiple weak learners, GBM is less likely to overfit compared to a single complex model.

### Real-World Case Study

**Netflix** uses GBM for its recommendation system. By iteratively improving predictions based on user interactions, Netflix can provide highly personalized content recommendations, enhancing user experience and engagement.

### Hands-On Code Example



> **Try it in Google Colab:** [![Open in Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/shastrula/ailearningclub-courses/blob/main/ai-fundamentals/mod-15.ipynb)

```python
import numpy as np
from sklearn.ensemble import GradientBoostingRegressor

# Sample data
X = np.array([[1, 2], [3, 4], [5, 6], [7, 8]])
y = np.array([10, 20, 30, 40])

# Initialize GBM
gbm = GradientBoostingRegressor(n_estimators=100, learning_rate=0.1, max_depth=3)

# Fit the model
gbm.fit(X, y)

# Predict
predictions = gbm.predict(X)
print(predictions)
```

## Feature Engineering Techniques

### In-depth Explanation

**Feature engineering** involves creating new features or transforming existing ones to improve model performance. Effective feature engineering can significantly enhance the predictive power of machine learning models.

### Techniques

1. **Polynomial Features**: Create new features by raising existing features to different powers.
2. **Interaction Terms**: Generate new features by multiplying two or more existing features.
3. **Domain-Specific Transformations**: Apply transformations based on domain knowledge (e.g., log transformation for skewed data).

### Why Feature Engineering?

- **Improved Model Performance**: New features can capture complex relationships in the data that simple features might miss.
- **Dimensionality Reduction**: Transforming features can sometimes reduce the dimensionality of the dataset, making models more efficient.

### Real-World Case Study

**Kaggle Competitions** often see top performers using advanced feature engineering techniques. For example, in the **House Prices: Advanced Regression Techniques** competition, winners created numerous engineered features to achieve high predictive accuracy.

### Hands-On Code Example



> **💡 Tip:** When performing feature engineering, always evaluate the impact of new features on model performance using cross-validation to avoid overfitting.

```python
import pandas as pd
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

# Sample data
df = pd.DataFrame({'feature1': [1, 2, 3, 4], 'feature2': [5, 6, 7, 8]})

y = pd.Series([10, 20, 30, 40])

# Polynomial features
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(df)

# Linear regression
model = LinearRegression()
model.fit(X_poly, y)

# Predict
predictions = model.predict(X_poly)
print(predictions)
```
